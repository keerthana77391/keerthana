```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
x= np.array([
     [1,30],
     [2,45],
     [3,50],
     [4,40],
     [5,85],
     [6,80],
     [7,70],
     [8,90]],dtype=np.float32)
y=np.array([0,0,0,0,1,1,1,1],dtype=np.float32)
model=models.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(4,activation='relu'),
    layers.Dense(1,activation='sigmoid')
])
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.05),loss='binary_crossentropy',metrics=['accuracy'])
model.fit(x,y,epochs=100,verbose=0)
x_test = np.array([
    [2,35],
    [3,85],
    [7,85],
    [1,20]
],dtype=np.float32)
test_preds=model.predict(x_test)
for features,prob in zip(x_test,test_preds):
  result= "Pass" if prob[0]>0.5 else "Fail"
  print(f"Hours: {features[0]},Attendance:{features[1]}% -> Predicted: {result} ({prob[0]:.2%})")
```
